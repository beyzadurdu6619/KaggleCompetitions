```python
train=pd.read_csv("/kaggle/input/competitions/cs-755-regression/train.csv")
test=pd.read_csv("/kaggle/input/competitions/cs-755-regression/test.csv")
```

```python
train.info()
```

```python
y=train["temp"]
x=train.drop(columns=["temp"],axis=1)
```

```python
test
```

```python
y = y.fillna(y.mean())
```

```python
import lightgbm as lgb
from sklearn.model_selection import train_test_split
lr=lgb.LGBMRegressor(
    n_estimators=150,  # Ağaç sayısını dengeli bir seviyeye çektik
    learning_rate=0.05,
    max_depth=5,  # Ağacın çok derinleşip ezberlemesini engelledik
    random_state=42,
    verbose=-1,
)
y_train_log = np.log1p(y)
model=lr.fit(x,y_train_log)
test_x=test.copy().drop(columns=["id"],axis=1)
tahminler=model.predict(test_x)
df = pd.DataFrame({"id": test["id"], "temp": tahminler})
df.head()
```

```python
df.to_csv("submission.csv", index=False)
```
